```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
# load the data
df= sns.load_dataset('iris')
print("First 5 rows.")
print(df.head())
print("\nDataset size:",df.shape)
```

```python
#See Flower Types
print("Types of Flowers:")
print(df['species'].value_counts())
```

```python
# BLOCK 3: Draw scatter plot (shows relationship)
plt.figure(figsize=(8,5))
sns.scatterplot(data=df, x='petal_length', y='petal_width', hue='species')
plt.title('Flower Measurements - Different Colors = Different Flowers')
plt.show()
```

```python
# BLOCK 4: Draw histogram (shows distribution)
plt.figure(figsize=(8,5))
plt.hist(df['sepal_length'], bins=15, color='lightblue', edgecolor='black')
plt.title('Distribution of Sepal Length')
plt.xlabel('Length (cm)')
plt.ylabel('Number of Flowers')
plt.show()
```

```python
# BLOCK 5: Draw box plot (shows outliers)
plt.figure(figsize=(8,5))
sns.boxplot(data=df, x='species', y='sepal_width')
plt.title('Sepal Width Box Plot')
plt.show()
```

```python
# BLOCK 6: Quick stats
print("\n📈 Average measurements:")
print(df.groupby('species').mean())
```

```python
# BLOCK 7: Save your conclusion
print("\n" + "="*50)
print("MY CONCLUSION:")
print("="*50)
print("1. Three flower types: Setosa, Versicolor, Virginica")
print("2. Setosa flowers have the smallest petals")
print("3. Virginica flowers have the largest petals")
print("4. No missing data - clean dataset")
print("5. Petal length and width are related")
```
